# Number
In Python, a number can be an **integer** or a **floating-point (float)** number.

````python
# Example of an integer number
i = 8
print("i:", i, "type:", type(i))
````

````python
# Example of a float number
f = 8.0
print("f:", f, "type:", type(f))
````

````python
# Example of a string contatining an integer number
s = "8.0"
print("s:", s, "type:", type(s))
````

With numbers we can perform **arithmetic operations**:  
```+``` : sum

```-``` : difference

```*``` : multiplication

```/``` : division

```//``` : integer division/floored quotient

```%``` : module/remainder

```**``` : exponent

````python
# Sum with integer numbers
a = 4
b = 5
c = a + b
print("c:", c, "type:", type(c))

# Sum with float numbers
a = 4.2
b = 5.5
c = a + b
print("c:", c, "type:", type(c))

# Sum with integer and float numbers --> the results is a FLOAT
a = 4
b = 5.5
c = a + b
print("c:", c, "type:", type(c))
````

````python
# Difference with integer numbers
a = 4
b = 5
c = a - b
print("c:", c, "type:", type(c))

# Difference with float numbers
a = 4.2
b = 5.5
c = a - b
print("c:", c, "type:", type(c))

# Difference with integer and float numbers --> the results is a FLOAT
a = 4
b = 5.5
c = a - b
print("c:", c, "type:", type(c))
````

````python
# Multiplication with integer numbers
a = 4
b = 5
c = a * b
print("c:", c, "type:", type(c))

# Multiplication with float numbers
a = 4.2
b = 5.5
c = a * b
print("c:", c, "type:", type(c))

# Multiplication with integer and float numbers --> the results is a FLOAT
a = 4
b = 5.0
c = a * b
print("c:", c, "type:", type(c))
````

````python
# Division with integer numbers --> the results is a FLOAT
a = 10
b = 2
c = a / b
print("c:", c, "type:", type(c))

# Division with float numbers
a = 4.2
b = 5.5
c = a / b
print("c:", c, "type:", type(c))

# Division with integer and float numbers --> the results is a FLOAT
a = 4
b = 5.5
c = a / b
print("c:", c, "type:", type(c))
````

````python
# Integer division with integer numbers
a = 10
b = 2
c = a // b
print("c:", c, "type:", type(c))

# Integer division with float numbers
a = 4.2
b = 5.5
c = a // b
print("c:", c, "type:", type(c))

# Integer division with integer and float numbers --> the results is a FLOAT
a = 4
b = 5.5
c = a // b
print("c:", c, "type:", type(c))
````

````python
# Module/remainder with integer numbers
a = 9
b = 2
c = a % b
print("c:", c, "type:", type(c))

# Module/remainder with float numbers
a = 4.2
b = 5.5
c = a % b
print("c:", c, "type:", type(c))

# Module/remainder with integer and float numbers --> the results is a FLOAT
a = 4
b = 5.5
c = a % b
print("c:", c, "type:", type(c))
````

````python
# Exponent with integer numbers
a = 9
b = 2
c = a ** b
print("c:", c, "type:", type(c))

# Exponent with float numbers
a = 4.2
b = 5.5
c = a ** b
print("c:", c, "type:", type(c))

# Exponent with integer and float numbers --> the results is a FLOAT
a = 4
b = 5.5
c = a ** b
print("c:", c, "type:", type(c))
````
